# Linux & Git Cheat Sheet.  

## Navigering

| Kommando | Beskrivning |
|-----------|-------------|
| `pwd` | Visa aktuell mapp |
| `ls` | Lista innehåll i aktuell mapp |
| `cd mappnamn` | Byt mapp |
| `cd ..` | Gå upp en nivå |
| `cd ~` | Gå till hemkatalogen |

## Skapa mappar och filer

| Kommando | Beskrivning |
|-----------|-------------|
| `mkdir mappnamn` | Skapa en mapp |
| `mkdir -p projekt/data-engineering` | Skapa flera mappar vid behov |
| `touch filnamn.txt` | Skapa en tom fil |

## VS Code

| Kommando | Beskrivning |
|-----------|-------------|
| `code .` | Öppna aktuell mapp i VS Code |

## System

| Kommando | Beskrivning |
|-----------|-------------|
| `sudo apt update` | Uppdatera paketlistan |
| `sudo apt upgrade -y` | Uppdatera installerade paket |
| `sudo apt install git -y` | Installera Git |

## Kontrollera installationer

| Kommando | Beskrivning |
|-----------|-------------|
| `git --version` | Visa Git-version |
| `python3 --version` | Visa Python-version |
| `which git` | Visa var Git är installerat |
| `which python3` | Visa var Python är installerat |

## Git

| Kommando | Beskrivning |
|-----------|-------------|
| `git clone <repo-url>` | Klona ett repo |
| `git status` | Visa status |
| `git add .` | Lägg till alla ändringar |
| `git commit -m "Meddelande"` | Skapa commit |
| `git push` | Skicka till GitHub |

## .gitignore

| Rad i `.gitignore` | Beskrivning |
|--------------------|-------------|
| `private/` | Ignorera mappen private |
| `.vscode/` | Ignorera VS Code-inställningar |
| `.env` | Ignorera miljövariabler och hemligheter |
